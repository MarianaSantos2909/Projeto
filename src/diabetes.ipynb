```python
def percentage_missing_values(df: pd.DataFrame, col: str) -> float:
    # Calcula a contagem de valores nulos na coluna específica
    missing_count = df[col].isnull().sum()
    
    # Calcula o total de linhas do DataFrame
    total_rows = len(df)
    
    # Calcula a porcentagem
    percentage = (missing_count / total_rows) * 100
    
    return percentage
```

#### Standardizing values

```python
def standardize_outcome(val):
    val = str(val).strip().lower()
    if val == '1' or val == 'positive':
        return 'Positive'
    elif val == '0' or val == 'negative':
        return 'Negative'
    else:
        return np.nan
    
def standardize_clinic_region(val):
    val = str(val).strip().lower()
    if val == 'north' or val == 'n':
        return 'North'
    elif val == 'south' or val == 'south-side':
        return 'South'
    else:
        return np.nan

def standardize_care_path(val):
    val = str(val).strip().lower()
    if val == 'routine follow-up' or val == 'routine':
        return 'Routine Follow-up'
    elif val == 'high-risk':
        return 'High-Risk'
    elif val == 'urgent monitoring':
        return 'Urgent Monitoring'
    else:
        return np.nan

def standardize_patient_segment(val):
    val = str(val).strip().lower()
    if val == 'adult':
        return 'Adult'
    elif val == 'senior':
        return 'Senior'
    elif val == 'mid-age':
        return 'Mid-Age'
    else:
        return np.nan
```

```python
# Applying functions
df['outcome'] = df['outcome'].apply(standardize_outcome)
df['clinic region'] = df['clinic region'].apply(standardize_clinic_region)
df['care path'] = df['care path'].apply(standardize_care_path)
df['patient segment'] = df['patient segment'].apply(standardize_patient_segment)
```

```python
def clean_numeric_pollution(df, columns):
    for col in columns:
    
        df[col] = df[col].astype(str).str.extract('(\d+)').astype(float)
        
        df[col] = df[col].astype('Int64')
    return df

df = clean_numeric_pollution(df, numerical_columns)
```

#  Missing value imputation

#### Impute missing values for numerical columns

```python
df[numerical_columns] = df[numerical_columns].fillna(df[numerical_columns].median())
```

#### Impute missing values for categorical columns

```python
# Verify whether missing values still exist
df.isnull().sum()
```

```python
def impute_categorical(df, columns):
    for col in columns:
        mode = df[col].mode()
        if len(mode) > 0:
            df[col] = df[col].fillna(mode[0])
        else:
            df[col] = df[col].fillna('unknown')
    return df

df = impute_categorical(df, categorical_columns)
```

### Verifying if there are **outliers** in numerical columns 

```python
def verify_outliers(df: pd.DataFrame, col: str) -> bool:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    outlier_report = {}

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
        
        
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_report[col] = len(outliers)
        
    if len(outliers) > 0:
        print(f"Coluna '{col}': {len(outliers)} outliers detectados (Limites: {lower_bound:.2f} a {upper_bound:.2f})")
            
    return outlier_report
```

```python
for coluna in numerical_columns:
    resultado = verify_outliers(df, coluna)
```
